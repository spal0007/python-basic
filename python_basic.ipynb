```python
# 5. Create a list and tuple in Python, and demonstrate how attempting to change an element works differently for each

# Creating a list and a tuple
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)

# Modifying an element in the list
my_list[1] = 99  # ✅ This works because lists are mutable
print("Modified list:", my_list)

# Attempting to modify an element in the tuple
try:
    my_tuple[1] = 99  # ❌ This will raise an error because tuples are immutable
except TypeError as e:
    print("Error while modifying tuple:", e)
```

```python
# 6. Write a function to demonstrate the behavior of mutable and immutable arguments.
# Function to demonstrate mutable and immutable arguments
def modify_data(num, my_list):
    num += 10             # Immutable (int): modification doesn't affect original
    my_list.append(100)   # Mutable (list): modification affects original
    print("Inside function:")
    print("Number (immutable):", num)
    print("List (mutable):", my_list)

# Initial values
number = 5
data_list = [1, 2, 3]

# Call the function
modify_data(number, data_list)

print("\nOutside function:")
print("Number (immutable):", number)       # Still 5
print("List (mutable):", data_list)        # Modified to [1, 2, 3, 100]
```

```python
#8. Write a program to demonstrate the use of logical operators
# Define some test variables
a = True
b = False

# Demonstrate logical operators
print("Logical Operators Demo:\n")

# AND operator
print("a and b:", a and b)     # False — both must be True

# OR operator
print("a or b:", a or b)       # True — at least one is True

# NOT operator
print("not a:", not a)         # False — reverses the value
print("not b:", not b)         # True
```

```python
#9. Write a Python program to convert user input from string to integer, float, and boolean types
# Get input from user
user_input = input("Enter something: ")

# Convert to integer
try:
    int_val = int(user_input)
    print("Integer conversion:", int_val)
except ValueError:
    print("❌ Cannot convert to integer.")

# Convert to float
try:
    float_val = float(user_input)
    print("Float conversion:", float_val)
except ValueError:
    print("❌ Cannot convert to float.")

# Convert to boolean
bool_val = bool(user_input)
print("Boolean conversion:", bool_val)
```

```python
#10. Write code to demonstrate type casting with list elements.
# Original list with string elements
string_list = ["10", "25.5", "", "0", "True"]

# Convert each element to different types
int_list = []
float_list = []
bool_list = []

for item in string_list:
    # Integer conversion
    try:
        int_list.append(int(float(item)))  # Convert via float first to handle decimal strings
    except ValueError:
        int_list.append("❌")

    # Float conversion
    try:
        float_list.append(float(item))
    except ValueError:
        float_list.append("❌")

    # Boolean conversion
    bool_list.append(bool(item))  # Empty string becomes False, others become True

# Display results
print("Original List:", string_list)
print("Integer Casts:", int_list)
print("Float Casts:  ", float_list)
print("Boolean Casts:", bool_list)
```

```python
#13. Write a Python program to find the sum of all even numbers between 1 and 50
# Initialize sum
even_sum = 0

# Loop through numbers from 1 to 50
for num in range(1, 51):
    if num % 2 == 0:
        even_sum += num

# Display the result
print("Sum of even numbers from 1 to 50 is:", even_sum)
```

```python
#14. Write a program to reverse a string using a while loop
# Take input from the user
original_str = input("Enter a string: ")

# Initialize variables
reversed_str = ""
index = len(original_str) - 1

# Reverse the string using a while loop
while index >= 0:
    reversed_str += original_str[index]
    index -= 1

# Display the result
print("Reversed string:", reversed_str)
```
